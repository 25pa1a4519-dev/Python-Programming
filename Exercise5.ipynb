```python
def subtraction(a,b,c):
  return a-b-c
res = subtraction(7,2,3)
print(res)
```

```python
def multiple_value(*num):
  s = sum(num)
  c = len(num)
  avg = s/c
  return s,c,avg
res_sum,res_len,res_avg = multiple_value(1,2,3,4,5)
print(f'The sum of numbers is {res_sum},the count is {res_len},the avg is {res_avg}')
```
