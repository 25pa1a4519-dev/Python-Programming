```python
#Write a python program to find number of duplicates in a list
a = list(map(int,input().split()))
s = set(a)
print(s)
print(len(a)-len(s))
```

```python
#Write a python program to find the how many numbers have duplicates in a list
l1 = [int(i) for i in input().split()]
l2 = []
for i in l1:
  if l1.count(i)>1:
    l2.append(i)
s = set(l2)
print(s)
print(len(s))
```

```python
#To print non repeated elements
l1 = [int(i) for i in input().split()]
l2 = []
for i in l1:
  if l1.count(i)>1:
    l2.append(i)
s1 = set(l2)
s2 = set(l1)
print(s2-s1)
```

```python
#Write a python program that prints missings numbers between 1 to n
n = int(input())
l = [int(i) for i in input().split()]
l1 = set(l)
print(l1)
elements = set(range(1,n+1))
print(elements)
print(elements-l1)
```
