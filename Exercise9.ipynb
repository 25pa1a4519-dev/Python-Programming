```python
#copying content of one file to another
with open('ram.txt','r') as f:
  data = f.read()
  print('Reading done')
with open('abhi.txt','w') as f1:
  f1.write(data)
  print('File copied')
```

```python
#program that prints each line in a file in reverse order
with open('ram.txt','r') as f:
  data = f.readlines()
  for i in data:
    print(i[::-1])
```

```python
#write a program to sort in a file
#file should have only lower case words
#must be lowered
with open('ram.txt','r') as f:
  data = f.read().split()
  data_lower = [i.lower() for i in data]
  sort_data = sorted(data_lower)
with open('ram2.txt','w') as f1:
  for i in sort_data:
    f1.write(i+'\n')
  print('done')
```
